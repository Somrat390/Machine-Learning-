# Linear Regression

```python
def predict_linear(X, w, b):
  return w * X + b
```

```python
import numpy as np

X = np.array([1,2,3,4,5])
y = np.array([5,7,4,6,1])
```

```python
w = 2
b = 1

prediction = predict_linear(X,w,b)
```

```python
def mae(y_test,y_pred):
    erro = np.abs(y_test-y_pred)
    print(erro)
    mae = np.sum(erro)/len(erro)
    print(mae)

```

```python
mae(y,y_pred)
```

```python
def mse(y_test,y_pred):
    erro = (y_test-y_pred)**2
    print(erro)
    mse = np.sum(erro)/len(erro)
    print(mse)

mse(y,y_pred)
```

# Gradient

```python
w = 0
b = 0
def error_with_predict(y,y_pred):
  return y - y_pred


learning_rate = 0.1

for epoch in range(1000):

    y_pred = predict_linear(X, w, b)

    error = error_with_predict(y,y_pred)



    dw = (-2*np.sum(X*error))/len(error)
    db = (-2*np.sum(error))/len(error)

    w = w - (learning_rate*dw)
    b = b - (learning_rate*db)

    if epoch % 100 == 0:
      mse = np.mean(error ** 2)

      print("epoch:", epoch)
      print("w:", w)
      print("b:", b)
      print("MSE:", mse)
```
